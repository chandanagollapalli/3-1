```python
import xml.etree.ElementTree as ET
import networkx as nx

# Parse XML
tree = ET.parse("webpages.xml")
root = tree.getroot()

# Create Directed Graph
G = nx.DiGraph()

# Store page topics
topics = {}

# Read XML data
for page in root.findall("page"):
    page_id = page.get("id")
    topic = page.get("topic")

    topics[page_id] = topic
    G.add_node(page_id)

    for link in page.findall("link"):
        G.add_edge(page_id, link.text)

# Display Graph Information
print("Nodes:", G.nodes())
print("Edges:", G.edges())

# Topic for ranking
selected_topic = "AI"

# Personalization Vector
personalization = {}

for node in G.nodes():
    if topics.get(node) == selected_topic:
        personalization[node] = 1
    else:
        personalization[node] = 0

# Normalize
total = sum(personalization.values())
for node in personalization:
    personalization[node] /= total

# Compute Topic-Specific PageRank
pagerank = nx.pagerank(
    G,
    alpha=0.85,
    personalization=personalization
)

print("\nTopic-Specific PageRank (Topic = AI)\n")

for page, rank in sorted(
        pagerank.items(),
        key=lambda x: x[1],
        reverse=True):
    print(page, ":", round(rank, 4))
```
